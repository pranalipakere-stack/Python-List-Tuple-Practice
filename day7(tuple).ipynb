```python
#Q6.Find the largest element in a tuple.
T4=(20,40,60,90,100)
largest=T4[0]
for i in T4:
    if i > largest:
        largest=i
print(largest)
```

```python
#Q7.Find the smallest element in a tuple.
smallest=T4[0]
for i in T4:
    if i < smallest:
        smallest=i
print(smallest)        
```

```python
#Q8.Find the sum of all elements in a tuple.
sum=0
for i in T4:
    sum=sum+i
print(sum)       
```

```python
#Q9.Count the number of even elements in a tuple.
count=0
for i in T4:
    if(i%2==0):
        count=count+1
print(count)        
```

```python
#Q10.Search for a number in a tuple and print whether it is found or not.
num=int(input("enter a number"))
if num in T4:
    print("found")
else:
    print("not found")
```
