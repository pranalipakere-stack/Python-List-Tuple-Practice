```python
#Q!.Create a list of 5 numbers and print all elements.
L1=[10,20,30,40,50]
print(L1)
```

```python
#Q2.Take 5 numbers from the user and store them in a list.
num=int(input("enter a list of number :"))
num1=int(input("enter a list of number :"))
num2=int(input("enter a list of number :"))
num3=int(input("enter a list of number :"))
num4=int(input("enter a list of number :"))
L2=[num,num1,num2,num3,num4]
print(L2)
```

```python
#Q3.Find the largest element in a list.
largest=L1[0]
for i in L1:
    if i > largest:
        largest=i
print(largest)
```

```python
#Q4.Find the smallest element in a list.
smallest=L1[0]
for i in L1:
    if i < smallest:
        smallest=i
print(smallest)        
```

```python
#Q5.Find the sum of all elements in a list.
sum=0
for i in L1:
    sum=sum+i
print(sum)
```

```python
#Q6.Count the number of even elements in a list.
count=0
for i in L1:
    if(i%2==0):
        count=count+1
print(count)        
```

```python
#Q7.Count the number of odd elements in a list.
count=0
for i in L1:
    if(i%2!=0):
        count=count+1
print(count)        
```

```python
#Q8.Search for a number in a list and print whether it is found or not.
num=int(input("enter a number"))
if num in L1:
    print("found")
else:
    print("not found")
```

```python
#Q9.Print all elements of a list in reverse order.
for i in L1[::-1]:
    print(i)
```

```python
#Q10.Find the average of all numbers in a list.
sum=0
for i in L1:
    sum=sum+i
average=sum/len(L1)
print(average)
```
